```python
import numpy as np
import matplotlib.pyplot as plt
import h5py
import matplotlib.patches as patches

fontsize = 7
plt.rcParams.update({
    'font.family': 'Arial',     # or 'Helvetica' if embedded properly
    'font.size': fontsize,
    'pdf.fonttype': 42,         # Embed TrueType (for editable text)
    'ps.fonttype': 42,
    'axes.labelsize': fontsize,
    'xtick.labelsize': fontsize,
    'ytick.labelsize': fontsize,
    'axes.facecolor': 'None',   # transparent background if needed
    'axes.edgecolor': 'black',
    'axes.linewidth': 0.5
})

blue = '#1E88E5'
red = '#FF1D25'
```

```python
dir0 = "../"
dir1 = dir0+'data/'
dirf = dir0+'figures/'
```

## Fig. 1A

```python

aplot = 10 + np.arange(21)
fig, ax = plt.subplots( figsize=(2,2))
ax.set_xlabel(r'$a_{fs}$')
ax.set_ylabel(r'$a_{sf}$')
ax.set_xlabel('Advantage of the Fast, 'r'$a_{fs}$')
ax.set_ylabel('Advantage of the Slow, 'r'$a_{sf}$')
ax.plot( np.zeros(10), np.linspace(-1, 1, 10),  'k', alpha=0.5, linewidth=0.5, linestyle='--' )
ax.plot( np.linspace(-1, 1, 10), np.zeros(10),  'k', alpha=0.5, linewidth=0.5, linestyle='--' )

ax.set_yticks([0, 0.2]); ax.set_yticklabels([0, r'$\Delta r$'])
ax.set_xticks([-0.2, 0]); ax.set_xticklabels([r'$-\Delta r$', 0])

# Create a rectangle: (x, y), width, height
rect = patches.Rectangle((-0.6, 0.2), 0.4, 0.4, edgecolor='none', facecolor=blue, alpha=0.6)
ax.add_patch(rect)
rect = patches.Rectangle((-0.6, -0.6), 0.4, 0.8, edgecolor='none', facecolor='grey', alpha=0.6)
ax.add_patch(rect)
rect = patches.Rectangle((-0.2, -0.6), 0.8, 0.8, edgecolor='none', facecolor=red, alpha=0.6)
ax.add_patch(rect)
rect = patches.Rectangle((-0.2, 0.2), 0.8, 0.4, edgecolor='none', facecolor='violet', alpha=0.6)
ax.add_patch(rect)
ax.set_xlim(-0.6, 0.6)
ax.set_ylim(-0.6, 0.6)
fig.savefig(dirf + 'Fig1/1a_PD.pdf', bbox_inches='tight', transparent=True)
```

## Fig. 1c

```python
code = 'S1fraction'
date = "250501_L200_T300"
fname = "pd"

f = h5py.File(dir1 + code + '/'+ date + '/' + fname + '_hss.h5', "r"); hss = f["hss"][:]; f.close()
f = h5py.File(dir1 + code + '/'+ date + '/' + fname + '_fss.h5', "r"); fss = f["fss"][:]; f.close()
f = h5py.File(dir1 + code + '/'+ date + '/' + fname + '_ts.h5', "r"); ts = f["ts"][:]; f.close()

nbin_a = 41;
a12s = np.linspace(-2, 2, nbin_a)
a21s = np.linspace(-2, 2, nbin_a)
r1 = 1.0
r2 = 0.8

hss = hss.swapaxes(3,0); hss = hss.swapaxes(2,1)
fss = fss.swapaxes(3,0); fss = fss.swapaxes(2,1)
xs = np.linspace(0, 1, 201)
```

```python
f1 = 0.5; f2 = 0.5
i= 15; j = 25

hs = hss[i,j]; fs=fss[i,j]; 
fig, a = plt.subplots(figsize=(1.2,0.9))
a.set_xticks([]); a.set_yticks([])

rect = patches.Rectangle((0,0), 1,100, color='w', alpha=1)
a.add_patch(rect)
nl = 100
for tt in np.arange(nl):
    t = tt+1
    h = hs[:101,t]; f = fs[:101,t]; 
    a.plot( xs[:101][f>f2], h[f>f2], color=red, alpha=1)
    a.plot( xs[:101][f<f1], h[f<f1], color=blue, alpha=1)
    a.plot( 1-xs[:101][f>f2], h[f>f2], color=red, alpha=1)
    a.plot( 1-xs[:101][f<f1], h[f<f1], color=blue, alpha=1)
a.set_xlim(0.25, 0.75)
a.set_ylim(10,85)
fig.savefig(dirf + 'Fig1/1c_a.pdf', bbox_inches='tight', transparent=True)
```
